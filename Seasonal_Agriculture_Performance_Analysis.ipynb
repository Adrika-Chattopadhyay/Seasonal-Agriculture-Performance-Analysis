#### Imputing all the missing values with Median value of Rainfall_mm Column

---



```python
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
```

For **'Rainfall_mm'** column **mean** is **601.15** and **median(50%)** is **582.0** and **skewness score** is very low **0.139543**.

A skewness score between -0.5 and 0.5 indicates that this specific column's data is **highly symmetrical** and follows a normal-like distribution. However, because rainfall naturally relies on seasonal conditions, median is used for imputing the missing values of the column.  

#### Imputing all the missing values with Mean value of Soil_Moisture_pct Column

---



```python
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].mean())
```

For **'Soil_Moisture_pct'** column **mean** is **26.50** and **median(50%)** is **26.40** and **skewness score** is an almost perfect **-0.031709**.

This column is **perfectly symmetrical** (skewness is nearly zero). Therefore, mean is used for imputing the missing values of the column because it utilizes information from every single data point in the calculation.  

#### Imputing all the missing values with Median value of Yield_Tonnes_Ha Column

---



```python
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())
```

For **'Yield_Tonnes_Ha'** column **mean** is **5.27** but **median(50%)** is only **1.74** and **skewness score** is massive **4.039389**.

A skewness score greater than 1.0 indicates extreme right-hand skewness. Therefore, median is used for imputing the missing values of the column as it ensures the missing rows are filled with a value that genuinely represents a standard harvest.  

### Classification of Variables present in the Dataset


---



#### 1. Identifier Variables
These uniquely identify each record but typically don't have direct analytical value for modeling, though they can be useful for joining or tracking.
*   **Farm_ID:** Unique identifier for each farm.

#### 2. Categorical Variables
These represent types, groups, or classifications and are crucial for understanding variations across different conditions or segments.
*   **State:** Geographical region.
*   **District:** Sub-region within a state.
*   **Crop:** Type of crop cultivated.
*   **Season:** Agricultural season (e.g., Kharif, Rabi, Zaid). **Highly relevant to seasonal performance.**
*   **Irrigation_Method:** Type of irrigation used.

#### 3. Numerical Variables
These are quantitative measures and form the basis for most analytical models.

#### 4. Environmental Variables
These describe the natural conditions affecting agriculture. They are **highly relevant to seasonal performance** as seasons are defined by climatic patterns.
*  **Rainfall_mm:** Total rainfall in millimeters.
*   **Avg_Temperature_C:** Average temperature in Celsius.
*   **Humidity_pct:** Percentage of humidity.
*   **Sunlight_Hours_Day:** Average hours of sunlight per day.
*   **Soil_pH:** Soil acidity/alkalinity.
*   **Soil_Moisture_pct:** Percentage of soil moisture.
*   **Disease_Pest_Risk_pct:** Percentage risk of disease/pests (often influenced by environmental factors).

#### 5. Operational Variables
These describe the practices and inputs used in farming.
*   **Farm_Area_Hectares:** Size of the farm area in hectares.
*   **Nitrogen_kg_ha:** Nitrogen applied per hectare.
*   **Phosphorus_kg_ha:** Phosphorus applied per hectare.
*   **Potassium_kg_ha:** Potassium applied per hectare.
*   **Fertilizer_kg_ha:** Total fertilizer applied per hectare.
*   **Pesticide_Litre_ha:** Pesticide applied per hectare.
*   **Seed_Quality_Score:** Score indicating the quality of seeds.
*   **Water_Used_m3:** Total water used in cubic meters.
*   **Water_Efficiency_t_per_1000m3:** Water efficiency in tonnes per 1000 cubic meters.

#### 6. Production/Performance Variables
These are the direct outcomes of the agricultural process and are often target variables for prediction or optimization.
*   **Yield_Tonnes_Ha:** Crop yield in tonnes per hectare. **Highly relevant to seasonal performance.**
*   **Production_Tonnes:** Total production in tonnes. **Highly relevant to seasonal performance.**

#### 7. Economic Variables
These relate to the financial aspects of farming.
*   **Market_Price_INR_Tonne:** Market price per tonne in INR.
*   **Total_Cost_INR:** Total cost incurred in INR.
*   **Revenue_INR:** Total revenue generated in INR.
*   **Profit_INR:** Profit earned in INR.



### Variables Most Relevant to Seasonal Performance Problem


---





For understanding and modeling seasonal performance, the most relevant variables are:

*   **Season**: This is the primary categorical variable that directly defines the different agricultural periods, each with distinct environmental conditions and crop cycles.
*   **Environmental Variables (Rainfall_mm, Avg_Temperature_C, Humidity_pct, Sunlight_Hours_Day, Soil_pH, Soil_Moisture_pct, Disease_Pest_Risk_pct)**: These are crucial as seasonal changes are largely driven by variations in these climatic and soil conditions. They directly impact crop growth and potential risks.
*   **Production/Performance Variables (Yield_Tonnes_Ha, Production_Tonnes)**: These are typically the *dependent variables* or *target metrics* used to measure 'performance' in the context of seasonal agriculture. Understanding how they vary across seasons is central to the problem.
*   **Crop**: Different crops have different optimal growing seasons and environmental requirements. Analyzing performance across seasons *per crop* is essential.

```python
domain_rules = {
    'Farm_Area_Hectares': (0.1, 50.0),
    'Rainfall_mm': (0.0, 3000.0),
    'Avg_Temperature_C': (0.0, 55.0),
    'Humidity_pct': (0.0, 100.0),
    'Sunlight_Hours_Day': (0.0, 16.0),
    'Soil_pH': (3.5, 9.5),
    'Soil_Moisture_pct': (0.0, 100.0),
    'Nitrogen_kg_ha': (0.0, 500.0),
    'Phosphorus_kg_ha': (0.0, 500.0),
    'Potassium_kg_ha': (0.0, 500.0),
    'Fertilizer_kg_ha': (0.0, 1000.0),
    'Pesticide_Litre_ha': (0.0, 50.0),
    'Seed_Quality_Score': (0.0, 1.0),
    'Yield_Tonnes_Ha': (0.0, 150.0),
    'Production_Tonnes': (0.0, 10000.0),
    'Market_Price_INR_Tonne': (500, 500000),
    'Total_Cost_INR': (0, 5000000),
    'Revenue_INR': (0, 10000000),
    'Profit_INR': (-5000000, 10000000),
    'Water_Used_m3': (0, 500000),
    'Water_Efficiency_t_per_1000m3': (0.0, 200.0),
    'Disease_Pest_Risk_pct': (0.0, 100.0)
}

matrix_report = []

for col, (min_val, max_val) in domain_rules.items():
    if col in df.columns:
        col_data = pd.to_numeric(df[col], errors='coerce')

        actual_min = col_data.min()
        actual_max = col_data.max()


        out_of_bounds_lower = col_data[col_data < min_val].count()
        out_of_bounds_upper = col_data[col_data > max_val].count()
        total_violations = out_of_bounds_lower + out_of_bounds_upper

        matrix_report.append({
            'Column': col,
            'Expected Min': min_val,
            'Expected Max': max_val,
            'Actual Min': round(actual_min, 2) if not pd.isna(actual_min) else np.nan,
            'Actual Max': round(actual_max, 2) if not pd.isna(actual_max) else np.nan,
            'Action Required': "REMOVE ERRORS" if total_violations > 0 else "SAFE TO KEEP"
        })

domain_bound_matrix = pd.DataFrame(matrix_report)


print("\n\t\t\t\tOUTLIERS DOMAIN BOUND MATRIX ANALYSIS\n")
print(domain_bound_matrix.to_string(index=False))
```

Based on the visualizations and the calculated average values, here's how agricultural performance varies across the seasons:

Rabi Season: The Rabi season comes in second, with an average yield of 5.04 Tonnes/Ha and an average production of 41.49 Tonnes. While slightly lower than Kharif, it still represents a significant portion of agricultural output.

Zaid Season: The Zaid season exhibits the lowest average performance, with an average yield of 4.64 Tonnes/Ha and an average production of 38.89 Tonnes. This suggests it is typically the least productive season.

In conclusion, Kharif is the most productive season in terms of both yield and production, followed by Rabi, with Zaid showing the lowest performance.

   Characterised by the highest rainfall levels (often exceeding 600mm–1000mm) and maximum humidity pct. Crops like Rice, Cotton, and Maize dominate this window due to the heavy water requirement.
   
* **Rabi (Winter Season):**
  
  Defined by lower average temperatures, moderate to low rainfall, and lower humidity pct. This season sees high cultivation of Wheat, Pulses, and Sugarcane.


* **Zaid (Summer Season):**
  
  Marked by the highest average temperatures (often crossing 30°C–35°C), maximum sunlight hours per day, and low natural rainfall, relying heavily on artificial irrigation like Drip or Sprinkler systems. Short-duration variants of Maize and Pulses are common.

* **Financial Trends:**
  Sugarcane and Chilli crops generally yield the highest revenue and profit margins across their respective seasons, while resource-heavy setups with flood irrigation face variable profit lines.

### Suggestions for Better Seasonal Agricultural Planning

---



#### 1. Optimizing Crop Selection Based on Environmental Patterns:

* **Kharif (Monsoon):** Given its high rainfall, humidity, and disease/pest risk, planning should prioritize water-intensive crops (like Rice, Sugarcane) that thrive in these conditions, and incorporate robust pest/disease management strategies. Farmers should also consider drought-resistant varieties for areas prone to erratic monsoon patterns.

* **Rabi (Winter):** With moderate rainfall, cooler temperatures, and lower humidity/pest risk, Rabi is ideal for temperate crops (Wheat, Pulses). Planning should focus on efficient water use where rainfall is insufficient and leveraging the reduced pest pressure.

* **Zaid (Summer):** The hot, dry conditions with high sunlight hours but low natural water availability make Zaid challenging. Planning must focus on heat-tolerant and short-duration crops (specific Maize/Pulses varieties), coupled with advanced irrigation techniques (Drip/Sprinkler) and efficient water management. The significant financial losses in Zaid highlight the need for careful selection of high-value crops that can offset irrigation costs, or for farmers to explore alternative income streams during this period.

#### 2. Enhancing Resource Management:

* **Water Usage & Efficiency:** The finding that Zaid season has the highest water usage but lowest efficiency is critical. Planning should involve promoting water-saving irrigation methods (drip, sprinkler) in Zaid, potentially offering subsidies, and exploring crop varieties with lower water demands. In Kharif, while rainfall is high, efficient drainage and water harvesting can prevent waterlogging and conserve water for dry spells.

* **Fertilizer & Pesticide Application:** The consistent usage of fertilizers and pesticides across seasons, despite varying environmental conditions, suggests a potential for optimization. Planning can introduce soil testing and integrated pest management (IPM) practices to tailor inputs to specific seasonal and crop needs, reducing costs and environmental impact, especially during seasons with lower baseline risk.

#### 3. Improving Economic Viability:

* **Profit Maximization:** Kharif's high profitability confirms its strategic importance. Planning should aim to maximize output and market access during this season. For Rabi, focus on maintaining profitability through cost-effective practices. Zaid, being consistently unprofitable, requires a re-evaluation of farming practices or a shift to alternative, more resilient agricultural models (e.g., protected cultivation, high-value horticulture).

* **Risk Mitigation:** The higher disease/pest risk in Kharif necessitates investment in preventative measures and early detection systems. For Zaid, managing the financial risk associated with high costs and low yields is paramount; this could involve crop insurance, government support programs, or diversification into non-agricultural income.

#### 4. Leveraging State-Specific Deviations:

* **Regional Tailoring:** The unique patterns observed in states like Maharashtra (strong Rabi performance) and Andhra Pradesh (higher Zaid yields than Rabi) indicate that a one-size-fits-all approach is ineffective. Planning should be localized, leveraging these regional strengths. For Maharashtra, investing in Rabi crops could yield higher returns. For Andhra Pradesh, understanding the factors contributing to Zaid's yield strength, despite unprofitability, could inform strategies to improve its economic outcomes.

* **Addressing Anomalies:** The presence of 'off-season' crops like Wheat in Kharif and Zaid in some regions points to potential misalignments between traditional practices and optimal conditions. Planning efforts should focus on educating farmers about optimal cropping calendars and promoting suitable varieties for each season, potentially through extension services or agricultural advisories.

### Key Insights from Seasonal Agriculture Performance Analysis

#### 1: Kharif Season Dominance in Productivity

4.  **Limitation:** While highly associated, this analysis doesn't establish causation between high rainfall and high humidity. Other meteorological factors influence both. Also, the data only reflects average values, and specific locations within a season may experience different conditions.

---



1.  **Observation:** The Zaid season exhibits the highest average temperatures and lowest rainfall, while Rabi has the lowest temperatures and moderate rainfall, and Kharif has moderate temperatures with the highest rainfall.

4.  **Limitation:** This shows association, not direct causation. Seasons are defined by these aggregate climatic conditions rather than one causing the other. Regional microclimates might deviate from these overall seasonal averages.

1.  **Observation:** The Kharif season, characterized by high humidity and rainfall, also shows the highest average disease and pest risk.

3.  **Interpretation:** The warm and humid conditions of the Kharif season create a more favorable environment for the proliferation of many agricultural pests and diseases, leading to higher observed risk percentages during this period.
4.  **Limitation:** While a strong association exists, high humidity and rainfall do not directly cause disease and pests; rather, they provide environmental conditions that facilitate their growth and spread. 4.  **Limitation:** This shows an association between high water usage and low efficiency in Zaid.
